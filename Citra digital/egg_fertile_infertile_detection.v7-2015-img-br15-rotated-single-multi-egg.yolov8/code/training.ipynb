```python
# =======================
# 🔹 1. Import library
# =======================

from ultralytics import YOLO
import pandas as pd
import matplotlib.pyplot as plt
import os
```

```python
# =======================
# 🔹 2. Path data.yaml
# =======================
data_yaml = r"C:/Asad/Kuliah/Penenlitian/pak latif/Citra digital/egg_fertile_infertile_detection.v7-2015-img-br15-rotated-single-multi-egg.yolov8/data.yaml"

# tampilkan isi data.yaml untuk verifikasi
with open(data_yaml, "r") as f:
    print(f.read())
```

```python
# =======================
# 🔹 3. Training YOLOv8
# =======================
# Gunakan model YOLOv8n (nano) dulu, ringan dan cocok untuk Raspberry Pi
model = YOLO("yolov8n.pt")

results = model.train(
    data=data_yaml,
    epochs=120,
    imgsz=320,
    batch=16,
    degrees=15,
    translate=0.1,
    scale=0.2,
    fliplr=0.5,
    hsv_h=0.015,
    hsv_s=0.7,
    hsv_v=0.4,
    mosaic=0.7,
    mixup=0.1,
    patience=30
)
```

```python
# =======================
# 🔹 4. Evaluasi di test set
# =======================
metrics = model.val(data=data_yaml, split="test", imgsz=320)
print(metrics)
```

```python
# =======================
# 🔹 5. Export model
# =======================
try:
    model.export(format="tflite", imgsz=320)
    print("✅ Export berhasil: TFLite")
except Exception as e:
    print("❌ Export TFLite gagal:", e)
    print("👉 Export ke ONNX saja")
    model.export(format="onnx", imgsz=320)
```

```python
# =======================
# 🔹 6. Visualisasi hasil training (versi fix)
# =======================
save_dir = model.trainer.save_dir
print("📂 Folder hasil training:", save_dir)

csv_path = os.path.join(save_dir, "results.csv")
df = pd.read_csv(csv_path)

# tampilkan daftar kolom biar tahu isinya
print("Kolom hasil:", df.columns.tolist())

# ambil nama kolom yang sesuai (tanpa spasi panjang)
col_precision = [c for c in df.columns if "precision" in c][0]
col_recall = [c for c in df.columns if "recall" in c][0]
col_map50 = [c for c in df.columns if "mAP50(" in c and "-95" not in c][0]
col_map5095 = [c for c in df.columns if "mAP50-95" in c][0]

# plot grafik
plt.figure(figsize=(8,5))
plt.plot(df[col_map5095], label="mAP50-95")
plt.plot(df[col_precision], label="Precision")
plt.plot(df[col_recall], label="Recall")
plt.xlabel("Epoch")
plt.ylabel("Score")
plt.title("Evaluasi YOLOv8")
plt.legend()
plt.grid()
plt.show()

# ringkasan untuk laporan
final = df.iloc[-1]
summary = {
    "Epoch": len(df),
    "Precision": final[col_precision],
    "Recall": final[col_recall],
    "mAP50": final[col_map50],
    "mAP50-95": final[col_map5095]
}
pd.DataFrame([summary])
```
